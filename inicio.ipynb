```python
import pandas as pd

tabela = pd.read_csv('exportacoes_franca.csv')
display(tabela)
display(tabela.info())

#Analise da Exportacoes da França ao longo dos anos
tabela_anos_exportacoes = tabela[["Year", "US$ FOB"]].groupby('Year').sum().reset_index()
def formatar_valor(valor):
    return f"US$ {valor:,.2f}"
tabela_anos_exportacoes["US$ FOB"] = tabela_anos_exportacoes["US$ FOB"].apply(formatar_valor)
display(tabela_anos_exportacoes)

#Produtos mais exportados pela França
tabela_produtos_exportados = tabela[["SH2 Description", "US$ FOB"]].groupby('SH2 Description').sum().reset_index().sort_values(by='US$ FOB', ascending=False)
tabela_produtos_exportados["US$ FOB"] = tabela_produtos_exportados["US$ FOB"].apply(formatar_valor)
display(tabela_produtos_exportados)

#Cidade que mais exportou em 2020
tabela_cidade_2020 = tabela[tabela["Year"] == 2020][["City", "US$ FOB"]].groupby('City').sum().reset_index().sort_values(by='US$ FOB', ascending=False)
tabela_cidade_2020["US$ FOB"] = tabela_cidade_2020["US$ FOB"].apply(formatar_valor)
display(tabela_cidade_2020)

#Quais produtos mais exportados cidades de 2020
cidade1 = tabela_cidade_2020.iloc[0]['City']
cidade2 = tabela_cidade_2020.iloc[1]['City']
tabela_cidade_2020 = tabela[(tabela["Year"] == 2020) & (tabela["City"].isin([cidade1, cidade2]))][["SH2 Description", "US$ FOB"]].groupby('SH2 Description').sum().reset_index().sort_values(by='US$ FOB', ascending=False)
tabela_cidade_2020["US$ FOB"] = tabela_cidade_2020["US$ FOB"].apply(formatar_valor)
display(tabela_cidade_2020)
print(f"Produtos mais exportados pela cidade {cidade1} em 2020 foi {tabela_cidade_2020.iloc[0]['SH2 Description']} com valor de {tabela_cidade_2020.iloc[0]['US$ FOB']}")
```
